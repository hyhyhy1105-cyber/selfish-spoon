## 문제 1. 은닉층 1개 모델 (25점)
`nn.Module` 을 상속해 아래 조건의 모델을 작성하세요.

- 입력 특성 5개, 은닉층 뉴런 16개, 출력 3개
- `__init__` 과 `forward` 를 모두 구현하고 각 줄에 주석
- `print(model)` 로 구조 출력


```python
# 답안
import torch
import torch.nn as nn

class Model1(nn.Module):
    def __init__(self):
        super(Model1, self).__init__()
        # 입력 특성 5개 -> 은닉층 뉴런 16개 선형 변환
        self.fc1 = nn.Linear(5, 16)
        # 활성화 함수 ReLU
        self.relu = nn.ReLU()
        # 은닉층 16개 -> 출력 3개 선형 변환
        self.fc2 = nn.Linear(16, 3)

    def forward(self, x):
        # 입력층에서 은닉층으로 전달
        out = self.fc1(x)
        # 활성화 함수 적용
        out = self.relu(out)
        # 은닉층에서 출력층으로 전달
        out = self.fc2(out)
        return out

model1 = Model1()
print(model1)
```

## 문제 2. 은닉층 2개 모델 (25점)
문제 1과 같은 입력·출력에 은닉층을 2개로 늘린 모델을 작성하고 구조를 출력하세요.


```python
# 답안
class Model2(nn.Module):
    def __init__(self):
        super(Model2, self).__init__()
        # 입력 특성 5개 ->첫 번째 은닉층 뉴런 16개
        self.fc1 = nn.Linear(5, 16)
        self.relu1 = nn.ReLU()
        # 첫 번째 은닉층 16개 -> 두 번째 은닉층 뉴런 16개
        self.fc2 = nn.Linear(16, 16)
        self.relu2 = nn.ReLU()
        # 두 번째 은닉층 16개 -> 출력 3개
        self.fc3 = nn.Linear(16, 3)

    def forward(self, x):
        out = self.fc1(x)
        out = self.relu1(out)
        out = self.fc2(out)
        out = self.relu2(out)
        out = self.fc3(out)
        return out

model2 = Model2()
print(model2)
```

## 문제 4. 구조 차이 설명 (20점)
아래 세 가지를 각각 두세 줄로 설명하세요.

1. 층을 늘리면 파라미터 수가 왜 늘어나는가
2. 활성화 함수가 없으면 층을 쌓아도 소용없는 이유
3. 층을 늘리는 것이 항상 좋은가 (근거를 함께)


**답:**
1. 층을 늘리면 파라미터 수가 왜 늘어나는가
선형 층(Linear layer)을 추가할 때마다 이전 층과 다음 층의 뉴런들을 연결하는 가중치(Weight)와 각 뉴런마다 더해지는 편향(Bias)이 새롭게 추가됩니다. 따라서 층을 쌓을수록 연결해야 하는 조합의 수가 곱해지면서 모델 내 학습 가능한 전체 파라미터 수가 대폭 증가하게 됩니다.
2. 활성화 함수가 없으면 층을 쌓아도 소용없는 이유
선형 연산(행렬 곱과 합)은 아무리 여러 번 중첩하더라도 단 하나의 선형 연산으로 합성 및 표현이 가능합니다 ($W_2(W_1 x + b_1) + b_2 = (W_2 W_1)x + (W_2 b_1 + b_2)$). 비선형 활성화 함수를 중간에 적용해주어야만 신경망이 비선형적인 복잡한 데이터 패턴과 결정 경계를 학습할 수 있는 표현력을 갖게 됩니다.
3. 층을 늘리는 것이 항상 좋은가 (근거를 함께)
층을 늘리는 것이 항상 좋은 것은 아닙니다. 층이 깊어질수록 학습 데이터에 과도하게 적합되어 새로운 데이터에 대한 성능이 떨어지는 오버피팅(Overfitting) 위험이 커지며, 역전파 과정에서 기울기가 사라지는 기울기 소멸(Gradient Vanishing) 문제나 연산량·메모리 증가로 인한 효율성 저하가 발생할 수 있습니다.
